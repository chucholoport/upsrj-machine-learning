# Importación de librerías

Este bloque importa las librerias numpy, pandas y time. Sin este bloque el resto del codigo fallará porque no habran sido incluidas las dependencias.

```python
import numpy as np
import pandas as pd
import time
```

```python
lista = list(range(100_000_000))

inicio = time.perf_counter()

suma = 0
for i in lista:
  suma += i

final = time.perf_counter()

tiempo = final - inicio

print(f"Tiempo de ejecucion: {tiempo:.6f} sec")
```

```python
np_lista = np.arange(100_000_000)

inicio = time.perf_counter()

suma_np = np.sum(np_lista)

final = time.perf_counter()

tiempo = final - inicio

print(f"Tiempo de ejecucion: {tiempo:.6f} sec")
```
