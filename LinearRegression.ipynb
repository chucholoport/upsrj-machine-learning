# Importacion de librerías

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

# Creación de datos de prueba

```python
# y = m * x + b

# Variable independiente
# SciKit-Learn requiere que se usen arreglos de 2D
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
print(X)

# Variable dependiente
y = np.array([2, 4, 5, 4, 5])
print(y)
```

# Graficación de datos originales

```python
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Datos originales")
plt.grid(visible=True)
plt.show()
```

# Generación y alimentación del modelo

```python
# Generamos nuestro modelo de regresión linear
model = LinearRegression()

# Alimentamos nuestro modelo
model.fit(X, y)

# Parámetros: m (peso), b (sesgo)
m = model.coef_[0]
b = model.intercept_

print(f"Pendiente (m): {m:.2f}")
print(f"Intercepto (b): {b:.2f}")
```

# Predicción y evaluación del modelo

```python
# Generamos una predicción
y_pred = model.predict(X)

# Evaluamos esa predicción
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Mean Squared Error: {mse:.2f}")
print(f"R2 Score: {r2:.2f}")
```

# Graficación del modelo

```python
plt.scatter(X, y, label="Datos reales")
plt.plot(X, y_pred, color="red", label="Modelo lineal")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Regresión linear simple")
plt.grid(visible=True)
plt.show()
```
