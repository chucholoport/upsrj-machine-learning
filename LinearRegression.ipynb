# Importacion de librerías

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

# Creación de datos de prueba

```python
# y = m * x + b

# Variable independiente
# SciKit-Learn requiere que se usen arreglos de 2D
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
print(X)

# Variable dependiente
y = np.array([2, 4, 5, 4, 5])
print(y)
```

# Graficación de datos originales

```python
plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Datos originales")
plt.grid(visible=True)
plt.show()
```

# Generación y alimentación del modelo

```python
# Generamos nuestro modelo de regresión linear
model = LinearRegression()

# Alimentamos nuestro modelo
model.fit(X, y)

# Parámetros: m (peso), b (sesgo)
m = model.coef_[0]
b = model.intercept_

print(f"Pendiente (m): {m:.2f}")
print(f"Intercepto (b): {b:.2f}")
```

# Predicción y evaluación del modelo

```python
# Generamos una predicción
y_pred = model.predict(X)

# Evaluamos esa predicción
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Mean Squared Error: {mse:.2f}")
print(f"R2 Score: {r2:.2f}")
```

# Graficación del modelo

```python
plt.scatter(X, y, label="Datos reales")
plt.plot(X, y_pred, color="red", label="Modelo lineal")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Regresión linear simple")
plt.grid(visible=True)
plt.show()
```

```python
# Generamos nuestro modelo de regresión linear
model = LinearRegression()

# Alimentamos nuestro modelo
model.fit(X, y)

# Parámetros: m (peso), b (sesgo)
m = model.coef_[0]
b = model.intercept_

print(f"Pendiente (m): {m:.2f}")
print(f"Intercepto (b): {b:.2f}")
```

```python
# Generamos una predicción
y_pred = model.predict(X)

# Evaluamos esa predicción
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print(f"Mean Squared Error: {mse:.2f}")
print(f"R2 Score: {r2:.2f}")
```

```python
plt.scatter(X, y, label="Datos reales")
plt.plot(X, y_pred, color="red", label="Modelo lineal")
plt.xlabel("Distancia al centro (km)")
plt.ylabel("Precio (miles de $)")
plt.title("Estimación de precio de una casa (regresión lineal simple)")
plt.grid(visible=True)
plt.show()
```

# Implementación de Pipeline

Un pipeline encapsula todo el flujo de procesamientos de datos en un solo objeto.

Para el usuario, entrenar un pipeline es igual que entrenar un modelo

El escalado no cambia el resultado conceptual, pero prepara el terreno para modelos más complejos

## Justificación

Hasta ahora hemos entrenado modelos paso por paso. En la vida real, esos pasos siempre se repiten en el mismo orden. Un pipeline es una forma de decirle a la computadora: **"Estos son los pasos, siempre ejecútalos así"**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
def plot_model(xlabel: str, ylabel: str, title: str):
  """
  Esta función grafica un modelo de regresión lineal.
  """
  plt.scatter(X, y, label="Datos reales")
  plt.plot(X, y_pred, color="red", label="Modelo lineal")
  plt.xlabel(xlabel)
  plt.ylabel(ylabel)
  plt.title(title)
  plt.grid(visible=True)
  plt.show()
```

```python
# 1. Definiremos las variables dependientes e independientes de nuestro modelo.

# Variable independiente X: Distancia al centro (km)
X = np.arange(1, 21).reshape(-1, 1)
print(X)

# Variable dependiente y: Precio (miles de $)
y = np.array([340, 320, 290, 260, 230,
              210, 220, 245, 275, 305,
              330, 350, 365, 375, 380,
              385, 390, 395, 398, 400])
print(y)
```

```python
# Definición del pipeline de regresión lineal
pipeline = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("regressor", LinearRegression())
])

# Entrenamiento del modelo
pipeline.fit(X, y)

# Predicción con el modelo
y_pred = pipeline.predict(X)

plot_model(xlabel="Distancia al centro (km)", ylabel="Precio (miles de $)",
           title="Estimación de precio de una casa (regresión lineal simple)")
```
